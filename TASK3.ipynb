```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# ==========================================
# 1. LOAD PREPROCESSED DATASET
# ==========================================
df = pd.read_csv('task 3 dataset.csv')
print("--- Dataset Loaded Successfully ---")
print(df.head())
```

```python
# ==========================================
# 2. DESCRIPTIVE STATISTICAL PROFILING
# ==========================================
print("\n--- Descriptive Statistics ---")
print(df.describe())

```

```python
# ==========================================
# 3. CORRELATION ANALYSIS & COST DRIVERS
# ==========================================
corr_columns = ['Distance (km)', 'Cargo Weight (kg)', 'Traffic Delay (mins)', 'Delivery Time (mins)']
corr_matrix = df[corr_columns].corr()
print("\n--- Correlation Matrix ---")
print(corr_matrix)

```

```python
# ==========================================
# 4. DATA VISUALIZATION (HEATMAP & SCATTER)
# ==========================================
# Correlation Heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap='Blues', fmt='.2f', linewidths=0.5)
plt.title('Correlation Heatmap - Logistics Operations', fontsize=12, fontweight='bold', color='#003366')
plt.tight_layout()
plt.savefig('task3_correlation_heatmap.png', dpi=300)
plt.show()

# Scatter Plot: Distance vs. Delivery Time colored by Traffic Delay
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Distance (km)', y='Delivery Time (mins)', hue='Traffic Delay (mins)', palette='viridis', s=60)
plt.title('Distance vs. Delivery Time (Colored by Traffic Delay)', fontsize=12, fontweight='bold', color='#003366')
plt.xlabel('Distance (km)')
plt.ylabel('Delivery Time (mins)')
plt.tight_layout()
plt.savefig('task3_scatter_plot.png', dpi=300)
plt.show()# ==========================================
# 2. DESCRIPTIVE STATISTICAL PROFILING
# ==========================================
print("\n--- Descriptive Statistics ---")
print(df.describe())

```
